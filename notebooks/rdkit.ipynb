```python
from rdkit import Chem
from rdkit.Chem import AllChem
from loguru import logger
```

```python
def get_Morgan(mol, radius=2, fpSize=1024):
    try:
        # 使用 GetMorganFingerprintAsBitVect 来生成指纹
        fp_generator = AllChem.GetMorganGenerator(radius=radius, fpSize=fpSize)
        fp = fp_generator.GetFingerprint(mol)
        return fp
    except Exception as e:
        logger.error(f"Morgan failed: {e}, using ECFP instead")
        return None

mol = Chem.MolFromSmiles('CCO')  # 生成分子对象
fp = get_Morgan(mol)  # 获取 Morgan 指纹

if fp is not None:
    print(fp)
else:
    print("Failed to generate Morgan fingerprint.")
```
